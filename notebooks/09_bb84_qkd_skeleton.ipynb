BB84 использует несовместимые базисы измерения. Если Bob выбирает тот же базис, что и Alice, бит может быть сохранён после sifting. Если базисы различаются, результат не входит в raw key. Именно эта логика позволяет обнаруживать вмешательство при оценке QBER (следующий проект — с Eve).
