# 08 — Shor's algorithm (факторизация 15)

Полная учебная цепочка по [IBM tutorial](https://quantum.cloud.ibm.com/docs/en/tutorials/shors-algorithm):
RSA → order finding → QPE → классическая пост-обработка.


```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))

from shor_utils import (
    shor_factorization_demo,
    create_order_finding_circuit_15,
    period_from_measurement,
)
from qiskit_aer import AerSimulator

demo = shor_factorization_demo(15, 2)
print(demo['rsa_note'])
print('Power table 2^x mod 15:', demo['power_table'][:8])
print('Period r =', demo['period'])
print('Factors:', demo['factors'])
```

```python
qc = create_order_finding_circuit_15(a=2, counting_qubits=5)
print('Order-finding circuit (QPE + modular mult mod 15):')
print(qc.draw('text'))

sim = AerSimulator()
counts = sim.run(qc, shots=2048).result().get_counts()
print('Sample counts:', dict(list(counts.items())[:5]))
r_est = period_from_measurement(counts, 5)
print('Estimated period (heuristic):', r_est)
```

## Связь с QFT

Quantum Phase Estimation извлекает фазу оператора $U|y\rangle=|2y\bmod 15\rangle$. Обратный QFT переводит фазу в измеримый битовый регистр.

## Ограничения (честно)

- Реальный RSA (2048+ бит) требует масштаба, недостижимого на NISQ.
- Схема модулярного умножения здесь учебная (N=15), не промышленная.
- Это **не** взлом RSA — демонстрация принципа.
