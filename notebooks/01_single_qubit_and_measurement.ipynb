# 01 — Один кубит и измерение


```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

sim = AerSimulator()
```

```python
def run(qc, shots=1024):
    return sim.run(qc, shots=shots).result().get_counts()

# |0>
qc0 = QuantumCircuit(1, 1)
qc0.measure(0, 0)
print('|0>:', run(qc0))

# |1>
qc1 = QuantumCircuit(1, 1)
qc1.x(0)
qc1.measure(0, 0)
print('|1>:', run(qc1))

# |+>
qc_plus = QuantumCircuit(1, 1)
qc_plus.h(0)
qc_plus.measure(0, 0)
counts = run(qc_plus, shots=4096)
print('|+>:', counts)
plot_histogram(counts)
plt.show()
```

**Вывод:** Hadamard gate переводит базисное состояние |0⟩ в суперпозицию. При измерении в computational basis результат становится вероятностным: при большом числе shots распределение стремится к 50% для 0 и 50% для 1.
