# 07 — Grover search demo


```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))

from circuits import create_grover_circuit, run_circuit
from visualization import plot_counts
import matplotlib.pyplot as plt

marked = '10'
before = run_circuit(create_grover_circuit(marked, iterations=0), shots=2048)
after = run_circuit(create_grover_circuit(marked, iterations=1), shots=2048)
print('Before:', before)
print('After:', after)
fig = plot_counts(after, 'Grover after 1 iteration')
fig.savefig(ROOT / 'images' / 'grover_histogram.png', bbox_inches='tight', dpi=150)
plt.show()
```

Grover не «угадывает ответ», а усиливает амплитуду целевого состояния (oracle + diffuser). Даёт квадратичное ускорение для неструктурированного поиска.
