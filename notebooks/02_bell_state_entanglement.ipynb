**Почему Bell state нельзя описать как два независимых случайных кубита?**

Совместное состояние |Φ+⟩ = (|00⟩+|11⟩)/√2 коррелирует результаты: только 00 и 11, почти нет 01 и 10. Это не произведение двух отдельных распределений.

Bell state — строительный блок телепортации, superdense coding и entanglement-based QKD.
